```python
import torch
import matplotlib.pyplot as plt
```

```python
history = torch.load("../models/history.pth")

train_losses = history["train_losses"]
train_accuracies = history["train_accuracies"]

print(train_losses)
print(train_accuracies)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(1, len(train_losses)+1),
    train_losses,
    marker="o",
    linewidth=2,
)

plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(1, len(train_accuracies)+1),
    train_accuracies,
    marker="o",
    linewidth=2,
)

plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.grid(True)

plt.show()
```
